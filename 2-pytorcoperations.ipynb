```python
import torch
import numpy as np
```

```python
#3 satır ve 5 süundan oluşan bir matriks oluşturur ve bu random sayılardan oluşur 0 ile 1 arasında
random_array=np.random.rand(3,5)
print(random_array)
```

```python
# 0 ile 10 arasında 3*5 matrix oluşturdu
random_matrix=torch.randint(10,size=(3,5))
print(random_matrix)
```

```python
zeros=torch.zeros(size=(3,2))
print(zeros)
```

```python
ones=torch.ones(size=(3,2))
print(ones)
```

```python
#kaçtan başlıcak, kaça kadar gidicek ve adım sayımız nedir. 
vector_with_arand=torch.arange(3,10,1) # 3 dahil ama 10 dahil değil olacak
print(vector_with_arand)
```

```python
similar_ones=torch.ones_like(input=vector_with_arand) #girdiğin tensoru 1 ile doldurarak yazıyor
print(similar_ones)
```
